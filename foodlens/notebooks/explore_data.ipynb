```python
import matplotlib.pyplot as plt
from pathlib import Path
from PIL import Image

RAW = Path("../data/raw")
dish = "jollof_rice"
files = sorted((RAW / dish).glob("*"))

print(f"Dish: {dish}")
print(f"Total images: {len(files)}")

fig, axes = plt.subplots(3, 4, figsize=(16, 12))
for ax, f in zip(axes.flat, files[:12]):
    img = Image.open(f)
    ax.imshow(img)
    ax.set_title(f.name)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
